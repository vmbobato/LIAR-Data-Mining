# Research Question and Hypotheses

## Research Question
How can combining text embeddings and speaker-graph embeddings improve detection of low-truth statements in LIAR, while controlling for speaker memorization bias?

## Hypotheses
1. A fused model (text + graph embeddings) outperforms text-only under fixed split evaluation.
2. Performance drops under speaker-disjoint evaluation (memorization pressure), but fused models retain a measurable advantage.
3. Stronger modeling (fine-tuned transformer, tuned tabular fusion, and optimized ensemble) improves low-truth detection over baseline linear models.
